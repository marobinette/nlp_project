```python
import regex_search
import pandas as pd

uvm_df = pd.read_excel('../data/uvm_ground_truth 1.xlsx')
```

```python
uvm_df['is_ug'] = uvm_df['cat_type'] == 'ug'
uvm_df['is_gr'] = uvm_df['cat_type'] == 'gr'
uvm_df['is_both'] = uvm_df['cat_type'] == 'both'

uvm_df['full_description'] = uvm_df['Title'] + ' ' + uvm_df['Description']
uvm_df['full_description'] = uvm_df['full_description'].astype(str)
uvm_df['full_description'] = uvm_df['full_description'].apply(lambda x: x.lower())
uvm_df.head()
```

```python
queries = [
    "thinking machines",
    "turing",
    "neural networks",
    "symbolic reasoning",
    "logic theorist",
    "cognitive revolution",
    "natural language processing",
    "natural language",
    "algorithm",
    "machine vision",
    "turing machine",
    "unsupervised learning",
    "a\.i\.",
    "semi-supervised learning",
    "reinforcement learning",
    "computational intelligence",
    "automata",
    "evolutionary computation",
    "bayesian networks"
]
obj = regex_search.KeyWordListSearchAgent(queries, courses_df=uvm_df)
```
